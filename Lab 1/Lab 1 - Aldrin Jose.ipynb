```python
def count_vowel(s):
  count = 0
  for char in s:
    if char in "aeiouAEIOU":
      count += 1
  return(count)
```

```python
count_vowel("Hello")
```

```python
def count_each_vowel(s):
    counts = {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}
    for char in s:
        if char.lower() in counts:
            counts[char.lower()] += 1
    return counts
```

```python
count_each_vowel("Hello")
```

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
```

```python
is_prime(9)
```

```python
def print_even_length_words(s):
    words = s.split()
    for word in words:
        if len(word) % 2 == 0:
            print(word)
```

```python
print_even_length_words("I am learning Python")
```

```python
def print_pattern(n):
    for i in range(1, n + 1):
        for j in range(i):
            print(i, end=" ")
        print()
```

```python
print_pattern(6)
```

```python
def is_symmetrical(s):
    mid = len(s) // 2
    first_half = s[:mid]
    second_half = s[mid:]
    return first_half == second_half
```

```python
is_symmetrical("Noon")
```

```python
def check_string(s):
    palindrome = s[::-1] == s
    mid = len(s) // 2
    symmetrical = s[:mid] == s[mid:]
    return palindrome, symmetrical
```

```python
check_string("noon")
```

```python
def reverse_string(s):
    return s[::-1]
```

```python
reverse_string("python")
```

```python
class Operations:
    def __init__(self, num1):
        self.num1 = num1

    def add(self, num2):
        return self.num1 + num2

    def subtract(self, num2):
        return self.num1 - num2

    def multiply(self, num2):
        return self.num1 * num2

    def divide(self, num2):
        return self.num1 / num2
```

```python
def split_even_odd(numbers):
    evens = []
    odds = []
    for num in numbers:
        if num % 2 == 0:
            evens.append(num)
        else:
            odds.append(num)
    return evens, odds
```

```python
split_even_odd([1, 2, 3, 4, 5, 6])
```

```python
def subject_with_max_marks(marks):
    best_subject = None
    best_mark = -1
    for subject, mark in marks.items():
        if mark > best_mark:
            best_subject = subject
            best_mark = mark
    return best_subject
```

```python
marks = {"math": 80, "english": 75, "science": 90, "computer": 95}
```

```python
subject_with_max_marks(marks)
```

```python
def max_profit(prices):
    min_price = prices[0]
    min_day = 0
    best_profit = 0
    buy_day = 0
    sell_day = 0
    for i, price in enumerate(prices[1:], start=1):
        profit = price - min_price
        if profit > best_profit:
            best_profit = profit
            buy_day = min_day
            sell_day = i
        if price < min_price:
            min_price = price
            min_day = i
    return best_profit, buy_day, sell_day
```

```python
prices = [7, 1, 5, 3, 6, 4]
max_profit(prices)
```

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0
    for num in num_set:
        if (num - 1) not in num_set:
            current = num
            length = 1
            while (current + 1) in num_set:
                current = current + 1
                length = length + 1
            if length > longest:
                longest = length
    return longest
```

```python
nums = [100, 4, 200, 1, 3, 2]
```

```python
longest_consecutive(nums)
```
